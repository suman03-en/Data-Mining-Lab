## 1. Write a program to find the following between two vectors a = [2,3,7,8,7] and b =[4,7,8,2,1].
    1. Euclidean distance
    2. Manhattan distance
    3. Cosine Similarity

```python
def euclidean_distance(point1, point2):
    """Calculate the Euclidean distance between two points in n-dimensional space."""
    
    array1 = np.array(point1)
    array2 = np.array(point2)
    inner_sum = np.sum((array1 - array2) ** 2)
    return np.sqrt(inner_sum)
```

```python
def manhattan_distance(point1, point2):
    array1 = np.array(point1)
    array2 = np.array(point2)
    abs_diff_term = np.abs((array1 - array2))
    return np.sum(abs_diff_term)
```

This function is used to calculate the magnitude of the arrays

```python
def magnitude(array1):
    return np.sqrt(np.sum(array1**2))
```

```python
def cosine_similarity(point1, point2):
    array1 = np.array(point1)
    array2 = np.array(point2)
    dot_product = np.dot(array1, array2)
    mag_arr1 = magnitude(array1)
    mag_arr2 = magnitude(array2)
    return dot_product / (mag_arr1 * mag_arr2)
```

```python
vector1 = [2, 3, 7, 8, 7]
vector2 = [4, 7, 8, 2, 1]

euclidean_distance_result = euclidean_distance(vector1, vector2)
manhattan_distance_result = manhattan_distance(vector1, vector2)
cosine_similarity_result = cosine_similarity(vector1, vector2)

print(f"euclidean_distance: {euclidean_distance_result:.2f}")
print(f"manhattan_distance: {manhattan_distance_result:.2f}")
print(f"cosine_similarity: {cosine_similarity_result:.2f}")
```

## 2. Write a program to find a simple matching coefficient and Jaccard similarity between [1,0,1,0,0,0] and [1,0,1,1,0,1].

```python
vector1 = np.array([1, 0, 1, 0, 0, 0])
vector2 = np.array([1, 0, 1, 1, 0, 1])

#calculating different coefficents using the logical operations on the vectors
m_0_0 = np.sum(np.logical_not(np.logical_or(vector1, vector2)))
m_0_1 = np.sum(np.logical_and(np.logical_not(vector1), vector2))
m_1_0 = np.sum(np.logical_and(vector1, np.logical_not(vector2)))
m_1_1 = np.sum(np.logical_and(vector1, vector2))

SMC = (m_0_0 + m_1_1) / (m_0_0 + m_0_1 + m_1_0 + m_1_1)
Jaccard_similarity = m_1_1 / (m_0_1 + m_1_0 + m_1_1)

print(f"m_0_0: {m_0_0}")
print(f"m_0_1: {m_0_1}")
print(f"m_1_0: {m_1_0}")
print(f"m_1_1: {m_1_1}")
print(f"SMC: {SMC:.3f}")
print(f"Jaccard_similarity: {Jaccard_similarity:.3f}")
```

## 3. Write a program to perform feature scaling (min max, z score and decimal scaling) on [35000, 42000, 39000, 45000]

3 functions for minmax, z_score and decimal scaling are implemented and A higher-level abstraction function, `scale_features` takes types of `scale_type` and return the scaled value.

```python
def minmax_scaling(data: NDArray) -> NDArray:
    """Scale features in the data using min-max scaling."""
    min_val = np.min(data, axis=0)
    max_val = np.max(data, axis=0)
    scaled_data = (data - min_val) / (max_val - min_val)
    return scaled_data

def z_score_scaling(data: NDArray) -> NDArray:
    """Scale features in the data using z-score scaling."""
    mean_val = np.mean(data, axis=0)
    std_dev = np.std(data, axis=0)
    scaled_data = (data - mean_val) / std_dev
    return scaled_data

def decimal_scaling(data: NDArray) -> NDArray:
    """Scale features in the data using decimal scaling."""
    max_val = np.max(data, axis=0)
    next_power_of_ten_of_max_val = 10 ** np.ceil(np.log10(max_val))
    scaled_data = data / next_power_of_ten_of_max_val
    return scaled_data
```

```python
SCALERS = {
    'minmax': minmax_scaling,
    'zscore': z_score_scaling,
    'decimal': decimal_scaling
}
def scale_features(data: NDArray, scale_type='minmax') -> NDArray:
    """Scale features in the data using specified scaling method.
    args:
        data: The input data to be scaled.
        scale_type: The type of scaling method to use ('minmax', 'zscore', 'decimal').
    returns:
        The scaled data.
    """
    try:
        return SCALERS[scale_type](data)
    except KeyError:
        raise ValueError(f"Invalid scale_type '{scale_type}'. Valid options are: {list(SCALERS.keys())}.")
```

```python
vector1 = np.array([35000, 42000, 39000, 45000])
scaled_vector1_minmax = scale_features(vector1, scale_type='minmax')
scaled_vector1_zscore = scale_features(vector1, scale_type='zscore')
scaled_vector1_decimal = scale_features(vector1, scale_type='decimal')

print(f"Original vector: {vector1}")
print(f"Scaled vector (Min-Max): {scaled_vector1_minmax}")
print(f"Scaled vector (Z-Score): {scaled_vector1_zscore}")
print(f"Scaled vector (Decimal Scaling): {scaled_vector1_decimal}")
```
