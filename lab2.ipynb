## 1. Write a program to find the following between two vectors a = [2,3,7,8,7] and b =[4,7,8,2,1].
    1. Euclidean distance
    2. Manhattan distance
    3. Cosine Similarity

```python
def euclidean_distance(point1, point2):
    """Calculate the Euclidean distance between two points in n-dimensional space."""
    
    array1 = np.array(point1)
    array2 = np.array(point2)
    inner_sum = np.sum((array1 - array2) ** 2)
    return np.sqrt(inner_sum)
```

```python
def manhattan_distance(point1, point2):
    array1 = np.array(point1)
    array2 = np.array(point2)
    abs_diff_term = np.abs((array1 - array2))
    return np.sum(abs_diff_term)
```

This function is used to calculate the magnitude of the arrays

```python
def magnitude(array1):
    return np.sqrt(np.sum(array1**2))
```

```python
def cosine_similarity(point1, point2):
    array1 = np.array(point1)
    array2 = np.array(point2)
    dot_product = np.dot(array1, array2)
    mag_arr1 = magnitude(array1)
    mag_arr2 = magnitude(array2)
    return dot_product / (mag_arr1 * mag_arr2)
```

```python
vector1 = [2, 3, 7, 8, 7]
vector2 = [4, 7, 8, 2, 1]

euclidean_distance_result = euclidean_distance(vector1, vector2)
manhattan_distance_result = manhattan_distance(vector1, vector2)
cosine_similarity_result = cosine_similarity(vector1, vector2)

print(f"euclidean_distance: {euclidean_distance_result:.2f}")
print(f"manhattan_distance: {manhattan_distance_result:.2f}")
print(f"cosine_similarity: {cosine_similarity_result:.2f}")
```

## 2. Write a program to find a simple matching coefficient and Jaccard similarity between [1,0,1,0,0,0] and [1,0,1,1,0,1].

```python
vector1 = np.array([1, 0, 1, 0, 0, 0])
vector2 = np.array([1, 0, 1, 1, 0, 1])

#calculating different coefficents using the logical operations on the vectors
m_0_0 = np.sum(np.logical_not(np.logical_or(vector1, vector2)))
m_0_1 = np.sum(np.logical_and(np.logical_not(vector1), vector2))
m_1_0 = np.sum(np.logical_and(vector1, np.logical_not(vector2)))
m_1_1 = np.sum(np.logical_and(vector1, vector2))

SMC = (m_0_0 + m_1_1) / (m_0_0 + m_0_1 + m_1_0 + m_1_1)
Jaccard_similarity = m_1_1 / (m_0_1 + m_1_0 + m_1_1)

print(f"m_0_0: {m_0_0}")
print(f"m_0_1: {m_0_1}")
print(f"m_1_0: {m_1_0}")
print(f"m_1_1: {m_1_1}")
print(f"SMC: {SMC:.3f}")
print(f"Jaccard_similarity: {Jaccard_similarity:.3f}")
```
