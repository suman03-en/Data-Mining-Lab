## 1. Write a program that takes three numbers and prints the largest one using if...elif...else. 

```python
numbers = []
max_num = None
print("Enter numbers: ")
for i in range(3):
    num = input(f"num{i} : ")
    try:
        numbers.append(int(num))
    except ValueError:
        print("Invalid input. Please enter a valid integer.")
for i in range(len(numbers)):
    if max_num is None or numbers[i] > max_num:
        max_num = numbers[i]
print(f"The numbers entered are: {numbers}")
print(f"The maximum number is: {max_num}")
```

## 2. Print the multiplication table of a given number using a for loop. 

```python
number = int(input("Enter number for multiplication table: "))
for i in range(1, 11):
    print(f"{number} x {i} = {number * i}")
```

## 3. Write a program that keeps asking the user to enter a number until they enter -1,  and then prints the sum of all entered numbers (excluding -1).

```python
numbers = []
sum = 0
while(True):
    num = int(input("Enter a number, or -1 to exit: "))
    if num == -1:
        break
    numbers.append(num)
    sum += num
print(f"The numbers entered are: {numbers}")
print(f"The sum of the numbers entered is: {sum}")
```

## 4. Given a list of tuples with student names and marks, write a program to print names of students who scored more than 80. 
students = [("Amit", 85), ("Rita", 78), ("John", 90)] 

```python
students = [("Amit", 85), ("Rita", 78), ("John", 90)]
print("Students with scores 80 or above:")
for name, score in students:
    if score >= 80:
        print(f"{name}: {score}")
```

## 5. Write a function is_prime(n) that returns True if a number is prime, else returns False. 

```python
def is_prime(n: int) -> bool:
    if n <= 1:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True
```

```python
is_prime(17)
```

## 6. Write a function is_palindrome(word) that returns True if the word is a palindrome (same forwards and backwards), and False otherwise.

```python
def is_palindrome(s: str) -> bool:
    return s == s[::-1]
```

```python
is_palindrome("racecar")
```
