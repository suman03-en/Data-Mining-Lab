1. Create a confusion matrix (accuracy ,precision, recall, specificity,fpr, ,f1score) from scratch without using sk-learn. (use numpy)

```python
import numpy as np 
import pandas as pd
import sklearn
from sklearn import tree
from sklearn import neighbors
from sklearn.model_selection import train_test_split
from sklearn.metrics import f1_score, precision_score, recall_score, accuracy_score, confusion_matrix
import matplotlib.pyplot as plt
```

```python
# [[TN, FP],
#  [FN, TP]]
confusion_matrix = np.array([[70, 10], 
                            [8, 50]])
print(confusion_matrix.shape)
```

```python
T_P = confusion_matrix[1, 1]
F_P = confusion_matrix[0, 1]
T_N = confusion_matrix[0, 0]
F_N = confusion_matrix[1, 0]

#precision
precision = T_P / (T_P + F_P)
accuracy = (T_P + T_N) / (T_P + T_N + F_P + F_N)
recall = T_P / (T_P + F_N)
specificity = T_N / (T_N + F_P)
fpr = 1 - specificity
f1_score = 2*precision*recall / (precision + recall)
print(f"Precision: {precision}")
print(f"accuracy: {accuracy}")
print(f"recall: {recall}")
print(f"specificity: {specificity}")
print(f"fpr: {fpr}")
print(f"f1_score: {f1_score}")


```

2. Construct a Decision Tree and KNN Algorithm on the Haberman dataset and evaluate its performance using accuracy,precision, recall and f1 score.

```python
# load the haberman dataset from the url, so internet is required
haberman_ds_url = "https://archive.ics.uci.edu/ml/machine-learning-databases/haberman/haberman.data"

columns = [
    "Age",
    "OperationYear",
    "PositiveNodes",
    "SurvivalStatus"
]

haberman_df = pd.read_csv(haberman_ds_url, names=columns)
print(f"Shape: {haberman_df.shape}")
print(haberman_df.head())

```

```python
#check the null values
print(haberman_df.isna().sum())
```

- Decision Tree

```python
y = haberman_df["SurvivalStatus"]
X = haberman_df.drop(columns="SurvivalStatus")
# split dataset to train and test
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=30, test_size=0.2)
```

### Using Decision Tree

```python
#constructing decision tree
clf = tree.DecisionTreeClassifier()
clf = clf.fit(X_train, y_train)
```

```python
# plot the decision tree
plt.figure(figsize=(15, 10), dpi=300)
tree.plot_tree(clf) # set the max_depth to small number to see the clear nodes
plt.title("Decision tree of haberman dataset", 
            fontdict= {'family': 'serif',
                        'color':  'darkred',
                        'weight': 'normal',
                        'size': 20,
                        })
plt.show()
```

```python
y_predicted = clf.predict(X_test)

# metrics
confusion_matrix_ds = confusion_matrix(y_test, y_predicted)

accuracy = accuracy_score(y_test, y_predicted)
precision = precision_score(y_test, y_predicted)
recall = recall_score(y_test, y_predicted)
f1_sc = f1_score(y_test, y_predicted)

print(confusion_matrix_ds)
print(f"accuracy: {accuracy}")
print(f"precision: {precision}")
print(f"recall: {recall}")
print(f"f1_sc: {f1_sc}")

```

### Using KNN Algorithm

```python
knn = neighbors.KNeighborsClassifier(n_neighbors=4)
knn = knn.fit(X_train, y_train)
y_pred = knn.predict(X_test)

#metrics 
confusion_matrix_knn = confusion_matrix(y_test, y_pred)
accuracy_knn = accuracy_score(y_test, y_pred)
precision_knn = precision_score(y_test, y_pred)
recall_knn = recall_score(y_test, y_pred)
f1_knn = f1_score(y_test, y_pred)
print(confusion_matrix_knn)
print(f"accuracy: {accuracy_knn}")
print(f"precision_knn: {precision_knn}")
print(f"recall_knn: {recall_knn}")
print(f"f1_knn: {f1_knn}")
```
