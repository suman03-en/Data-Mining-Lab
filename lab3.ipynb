# Lab 3: Pandas & Data Preprocessing

```python
import pandas as pd
```

```python
ecommerce_data = {
'OrderID': [101, 102, 103, 104, 105, 106, 107],
'CustomerName': ['Alice', 'Bob', 'Charlie', 'David', 'Eve', 'Frank', 'Grace'],
'Category': ['Electronics', 'Fashion', 'Electronics', 'Home', 'Fashion', 'Electronics', 'Home'],
'Amount': [200, 150, 300, 400, 250, 500, 350],
'Discount': [20, 10, 30, 50, 20, 60, 40],
'PurchaseDate': pd.to_datetime(['2023-11-20', '2023-11-21', '2023-11-22',
'2023-11-23', '2023-11-24', '2023-11-25', '2023-11-26']),
'City': ['New York', 'Los Angeles', 'New York', 'Chicago', 'Los Angeles', 'Chicago','New York']
}
```

```python
# Create the df from the given data: dict
df = pd.DataFrame(ecommerce_data)
```

1. What is the shape of the dataset?

```python
print(f"Shape of given data: {df.shape}")
print(df.columns)
```

```python
print(df.head())
```

2. Are there any null values in the dataset?

```python
print("Null values in dataset: ")
print(df.isna().sum())
```

3.Retrieve [Charlie Electronics 300] using loc and iloc

```python
#using iloc
data_1 = df.iloc[2,[1,2,3]]
data_2 = df.loc[2, ["CustomerName", "Category", "Amount"]]
print("Retrieved using iloc: ")
print(data_1)
print("\nRetrieved using loc: ")
print(data_2)
#print(data.to_dict().values())
```

4. Fetch rows for orders placed by customers from New York.

```python
q_filter = df["City"] == "New York"
data = df[q_filter]
print(data)
```

5. What is the average order amount? What is the total discount provided?

```python
avg_order_amt = df["Amount"].mean()
total_discount = df["Discount"].sum()
print(f"Average order amount: {avg_order_amt:.2f}")
print(f"Total discount provided: {total_discount}")
```

6. Sort the orders by Amount in descending order

```python
df.sort_values(by="Amount")
```
