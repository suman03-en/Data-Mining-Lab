## Question 2

```python
#exploring dataset
data_df = pd.DataFrame(data)
print(f"Shape: {data_df.shape}")
data_df.head()
```

```python
data_df.isna().sum()
```

1. Replace None in numerical fields (age, income) with mean/median.

```python
# imputing null values
age_mean = data_df["age"].mean()
income_mean = data_df["income"].mean()
fill_values = {"age": age_mean, "income": income_mean}
data_df = data_df.fillna(value=fill_values)
print(data_df)
```

2. Replace None in categorical fields (city, purchased) with mode or a placeholder
like "unknown".

```python
fill_values_cat = data_df[["city", "purchased"]].mode().iloc[0].to_dict()
data_df = data_df.fillna(value=fill_values_cat)
print(data_df)
```

3. Normalize numerical fields (age, income) to a scale of 0-1 using min-max scaling
or standardization (z-score).

```python
#zscore
data_df[["age", "income"]] = (
    (data_df[["age", "income"]] - data_df[["age", "income"]].mean())
    / data_df[["age", "income"]].std()
    ).round(3)
print(data_df)
```

4. Use one-hot encoding for the city

```python
encoded_data_df = pd.get_dummies(data_df, columns=["city"], dtype=int)
print(encoded_data_df)
```

5. Use label encoding (e.g., yes=1, no=0) for purchase.

```python
data_df["purchased"] = data_df["purchased"].map({"no": 0, "yes": 1})
print(data_df)
```
